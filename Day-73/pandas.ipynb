```python
reshaped_df = df.pivot(columns = "TAG", index = "Date", values = "POSTS")
reshaped_df
```

```python
reshaped_df.groupby(['assembly', 'c', 'c#', 'c++', 'delphi', 'go', 'java', 'javascript',
       'perl', 'php', 'python', 'r', 'ruby', 'swift']).count()
```

```python
reshaped_df.fillna(0 ,inplace = True)
```

```python
reshaped_df.isna().values.any()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python


plt.figure(figsize = (16,10))
plt.xlabel("Date", fontsize = 16)
plt.ylabel("Number of Ports", fontsize = 16)
plt.xticks(fontsize = 16)
plt.yticks(fontsize = 16)
plt.ylim(0,35000)


for column in reshaped_df.columns:
    plt.plot(reshaped_df.index,reshaped_df[column], linewidth = 4, label = column)
plt.legend(fontsize = 16)
```

```python
# The window is number of observations that are averaged
roll_df = reshaped_df.rolling(window = 12).mean()


plt.figure(figsize = (16,10))
plt.xlabel("Date", fontsize = 16)
plt.ylabel("Number of Ports", fontsize = 16)
plt.xticks(fontsize = 16)
plt.yticks(fontsize = 16)
plt.ylim(0,35000)

# plottint the roll_df instead
for column in roll_df.columns:
    plt.plot(reshaped_df.index,roll_df[column], linewidth = 4, label = column)
plt.legend(fontsize = 16)
```
